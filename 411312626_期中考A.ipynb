````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================

DB_NAME = "activity_system.db"

def init_db():
    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()

        # 開啟外鍵檢查
        cursor.execute("PRAGMA foreign_keys = ON")

        # 刪除舊資料表
        cursor.execute("DROP TABLE IF EXISTS registrations")
        cursor.execute("DROP TABLE IF EXISTS activities")

        # 建立活動資料表
        cursor.execute("""
            CREATE TABLE activities (
                activity_id INTEGER PRIMARY KEY,
                activity_name TEXT NOT NULL UNIQUE,
                activity_date TEXT NOT NULL,
                capacity INTEGER NOT NULL
            )
        """)

        # 建立報名紀錄資料表
        cursor.execute("""
            CREATE TABLE registrations (
                registration_id INTEGER PRIMARY KEY AUTOINCREMENT,
                student_name TEXT NOT NULL,
                activity_id INTEGER NOT NULL,
                people_count INTEGER NOT NULL DEFAULT 1,
                status TEXT NOT NULL DEFAULT '報名中',
                FOREIGN KEY (activity_id)
                    REFERENCES activities(activity_id)
            )
        """)

        # 三場活動的初始資料
        activities_data = [
            (1, "迎新晚會", "2026-11-20", 6),
            (2, "程式工作坊", "2026-11-25", 4),
            (3, "籃球友誼賽", "2026-11-28", 3)
        ]

        # 新增活動資料
        cursor.executemany("""
            INSERT INTO activities
            (activity_id, activity_name, activity_date, capacity)
            VALUES (?, ?, ?, ?)
        """, activities_data)

        conn.commit()

        print("✅ 資料庫建立成功！")
        print("✅ 已建立 activities 和 registrations 兩張資料表")
        print("✅ 已新增三場活動初始資料")

    finally:
        conn.close()


# 執行資料庫初始化
init_db()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================

# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================

# 允許的動作與活動
ALLOWED_ACTIONS = ["報名", "取消", "查詢", "統計", "未知"]
ALLOWED_ACTIVITIES = ["迎新晚會", "程式工作坊", "籃球友誼賽", "未知"]

# 暫存 AI 解讀結果，避免相同句子重複呼叫 API
ai_cache = {}


def parse_ai_response(raw):
    # 將 AI 回覆依照逗號分成三個欄位
    parts = re.split(r"[,，]", raw.strip())

    action = "未知"
    activity = "未知"
    people_count = 1

    # 檢查動作是否在允許清單中
    if len(parts) >= 1:
        value = parts[0].strip()
        if value in ALLOWED_ACTIONS:
            action = value

    # 檢查活動是否在允許清單中
    if len(parts) >= 2:
        value = parts[1].strip()
        if value in ALLOWED_ACTIVITIES:
            activity = value

    # 使用正規表示式找出人數
    if len(parts) >= 3:
        match = re.search(r"-?\d+", parts[2])
        if match:
            people_count = int(match.group())

    return action, activity, people_count


def ai_interpret(user_text):
    # 避免空白輸入
    user_text = str(user_text).strip()

    if not user_text:
        return ("未知", "未知", 1, "輸入為空白")

    # 相同句子直接使用暫存結果
    if user_text in ai_cache:
        return ai_cache[user_text]

    # 設定 AI 的工作規則
    system_prompt = """
你是系學會活動報名系統的語意解析助手。

你的工作只有從使用者的一句話中，
判斷「動作、活動、人數」。

【動作】
只能輸出：
報名、取消、查詢、統計、未知

【活動】
只能輸出：
迎新晚會、程式工作坊、籃球友誼賽、未知

【人數】
1. 沒有提到人數，輸出 1。
2. 「我和一個朋友」代表 2 人。
3. 「我和兩個朋友」代表 3 人。
4. 「我們三個人」代表 3 人。
5. 「總共兩人」代表 2 人。
6. 「總共四個人」代表 4 人。
7. 即使人數超過 3，也要照實輸出。
8. 不可以自行限制人數範圍。

【統計】
如果使用者說：
各活動報名狀況、所有活動報名統計、
全部活動報名情形

輸出：
統計,未知,1

【未知】
如果活動不是指定的三場活動，
活動輸出未知。

如果動作無法判斷，
動作輸出未知。

【輸出格式】
只能輸出一行：
動作,活動,人數

例如：
報名,迎新晚會,1
報名,籃球友誼賽,3
取消,程式工作坊,1
查詢,籃球友誼賽,1
統計,未知,1
報名,未知,1
未知,未知,1

不可以輸出解釋、Markdown 或其他文字。
"""

    # 最多嘗試 3 次
    for attempt in range(3):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {"role": "system", "content": system_prompt},
                    {"role": "user", "content": user_text}
                ],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )

            # 取得 AI 原始回覆
            raw = response.choices[0].message.content or ""

            # 不直接相信 AI，先驗證輸出
            action, activity, people_count = parse_ai_response(raw)

            result = (action, activity, people_count, raw)

            # 儲存暫存結果
            ai_cache[user_text] = result

            # 每次成功呼叫後等待 2 秒
            time.sleep(2)

            return result

        except Exception as e:
            # 不印出完整錯誤，避免意外洩漏敏感資訊
            error_text = str(e).lower()

            # 判斷是否為 429 使用量限制
            status_code = getattr(e, "status_code", None)

            if status_code == 429 or "429" in error_text:
                if attempt < 2:
                    print("⚠️ API 使用量限制，等待 30 秒後重試...")
                    time.sleep(30)
                    continue

                print("❌ API 使用量限制，已達最多重試次數")

            else:
                print("❌ AI 呼叫失敗，請檢查模型或 API 設定")

            return ("未知", "未知", 1, "AI 呼叫失敗")

    return ("未知", "未知", 1, "AI 呼叫失敗")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================

# 測試格式：
# (使用者說的話, 預期動作, 預期活動, 預期人數)

ai_test_cases = [
    ("我要報名迎新晚會", "報名", "迎新晚會", 1),
    ("幫我報名程式工作坊", "報名", "程式工作坊", 1),
    ("我要報名籃球友誼賽，我和兩個朋友一起", "報名", "籃球友誼賽", 3),
    ("我要報名迎新晚會，總共兩人", "報名", "迎新晚會", 2),
    ("幫我取消程式工作坊", "取消", "程式工作坊", 1),
    ("籃球友誼賽還有多少名額", "查詢", "籃球友誼賽", 1),
    ("各活動報名狀況", "統計", "未知", 1),
    ("我要報名演唱會", "報名", "未知", 1),
    ("我要報名迎新晚會，總共四個人", "報名", "迎新晚會", 4),
    ("今天天氣如何", "未知", "未知", 1)
]

# 記錄通過數量
passed = 0

print("========== AI 解讀測試開始 ==========")

# 依序執行 10 句測試
for i, (user_text, expected_action, expected_activity, expected_count) in enumerate(ai_test_cases, start=1):

    # 呼叫第 4 格的 AI 解讀函式
    action, activity, people_count, raw = ai_interpret(user_text)

    # 比對測試結果
    action_ok = (expected_action == "*" or action == expected_action)
    activity_ok = (expected_activity == "*" or activity == expected_activity)
    count_ok = (expected_count == "*" or people_count == expected_count)

    is_pass = action_ok and activity_ok and count_ok

    if is_pass:
        passed += 1

    # 印出測試結果
    print(f"\n===== 第 {i} 題 =====")
    print("使用者輸入：", user_text)
    print("AI 原始回覆：", raw)
    print("系統解讀：", (action, activity, people_count))
    print("預期結果：", (expected_action, expected_activity, expected_count))
    print("測試結果：", "PASS ✅" if is_pass else "FAIL ❌")

# 顯示總結
print("\n========== AI 解讀測試結束 ==========")
print(f"AI 解讀測試通過：{passed}/{len(ai_test_cases)}")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================

# 顯示目前有哪些活動
def available_activities():
    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()

        cursor.execute("""
            SELECT activity_name, activity_date, capacity
            FROM activities
            ORDER BY activity_id
        """)

        rows = cursor.fetchall()

        result = "目前有以下活動：\n"

        for name, date, capacity in rows:
            result += f"{name}（{date}），總名額 {capacity} 人\n"

        return result

    finally:
        conn.close()


# 查詢指定活動的剩餘名額
def get_remaining(activity):
    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()

        cursor.execute("""
            SELECT activity_id, capacity
            FROM activities
            WHERE activity_name = ?
        """, (activity,))

        row = cursor.fetchone()

        if row is None:
            return None

        activity_id, capacity = row

        cursor.execute("""
            SELECT COALESCE(SUM(people_count), 0)
            FROM registrations
            WHERE activity_id = ?
            AND status = ?
        """, (activity_id, "報名中"))

        count_row = cursor.fetchone()

        if count_row is None:
            registered = 0
        else:
            registered = count_row[0]

        return capacity - registered

    finally:
        conn.close()


# 報名功能
def register_activity(student_name, activity, people_count=1):

    student_name = str(student_name).strip()

    # 檢查姓名
    if not student_name:
        return "❌ 請輸入姓名"

    # 檢查人數
    if not isinstance(people_count, int) or not 1 <= people_count <= 3:
        return "❌ 每次只能報名 1～3 人"

    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()

        # 開啟外鍵檢查
        cursor.execute("PRAGMA foreign_keys = ON")

        # 開始交易，避免同時報名造成名額超收
        cursor.execute("BEGIN IMMEDIATE")

        # 查詢活動
        cursor.execute("""
            SELECT activity_id, capacity
            FROM activities
            WHERE activity_name = ?
        """, (activity,))

        row = cursor.fetchone()

        if row is None:
            return "❌ 找不到活動\n" + available_activities()

        activity_id, capacity = row

        # 檢查是否重複報名
        cursor.execute("""
            SELECT registration_id
            FROM registrations
            WHERE student_name = ?
            AND activity_id = ?
            AND status = ?
        """, (student_name, activity_id, "報名中"))

        existing = cursor.fetchone()

        if existing is not None:
            return "❌ 已報名此活動，不可重複報名"

        # 檢查每人最多報名兩場活動
        cursor.execute("""
            SELECT COUNT(DISTINCT activity_id)
            FROM registrations
            WHERE student_name = ?
            AND status = ?
        """, (student_name, "報名中"))

        count_row = cursor.fetchone()

        if count_row is None:
            activity_count = 0
        else:
            activity_count = count_row[0]

        if activity_count >= 2:
            return "❌ 每位學生最多同時報名兩個活動"

        # 計算目前已報名人數
        cursor.execute("""
            SELECT COALESCE(SUM(people_count), 0)
            FROM registrations
            WHERE activity_id = ?
            AND status = ?
        """, (activity_id, "報名中"))

        count_row = cursor.fetchone()

        if count_row is None:
            registered = 0
        else:
            registered = count_row[0]

        remaining = capacity - registered

        # 檢查名額是否足夠
        if remaining < people_count:
            return f"❌ 名額不足，目前只剩 {remaining} 人"

        # 檢查是否有已取消的舊紀錄
        cursor.execute("""
            SELECT registration_id
            FROM registrations
            WHERE student_name = ?
            AND activity_id = ?
            AND status = ?
            ORDER BY registration_id DESC
            LIMIT 1
        """, (student_name, activity_id, "已取消"))

        old_record = cursor.fetchone()

        if old_record is not None:
            # 取消後重新報名
            cursor.execute("""
                UPDATE registrations
                SET status = ?, people_count = ?
                WHERE registration_id = ?
            """, ("報名中", people_count, old_record[0]))

        else:
            # 新增報名紀錄
            cursor.execute("""
                INSERT INTO registrations
                (student_name, activity_id, people_count, status)
                VALUES (?, ?, ?, ?)
            """, (student_name, activity_id, people_count, "報名中"))

        conn.commit()

        new_remaining = remaining - people_count

        return (
            f"✅ 報名成功！\n"
            f"姓名：{student_name}\n"
            f"活動：{activity}\n"
            f"報名人數：{people_count} 人\n"
            f"剩餘名額：{new_remaining} 人"
        )

    finally:
        conn.close()


# 取消報名功能
def cancel_activity(student_name, activity):

    student_name = str(student_name).strip()

    if not student_name:
        return "❌ 請輸入姓名"

    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()

        cursor.execute("PRAGMA foreign_keys = ON")
        cursor.execute("BEGIN IMMEDIATE")

        # 查詢活動
        cursor.execute("""
            SELECT activity_id
            FROM activities
            WHERE activity_name = ?
        """, (activity,))

        row = cursor.fetchone()

        if row is None:
            return "❌ 找不到活動\n" + available_activities()

        activity_id = row[0]

        # 查詢有效報名紀錄
        cursor.execute("""
            SELECT registration_id, people_count
            FROM registrations
            WHERE student_name = ?
            AND activity_id = ?
            AND status = ?
        """, (student_name, activity_id, "報名中"))

        record = cursor.fetchone()

        if record is None:
            return "❌ 查無報名紀錄，無法取消"

        registration_id, people_count = record

        # 更新報名狀態
        cursor.execute("""
            UPDATE registrations
            SET status = ?
            WHERE registration_id = ?
        """, ("已取消", registration_id))

        conn.commit()

        remaining = get_remaining(activity)

        return (
            f"✅ 取消成功！\n"
            f"姓名：{student_name}\n"
            f"活動：{activity}\n"
            f"已釋出名額：{people_count} 人\n"
            f"目前剩餘名額：{remaining} 人"
        )

    finally:
        conn.close()


# 查詢活動剩餘名額
def query_activity(activity):

    remaining = get_remaining(activity)

    if remaining is None:
        return "❌ 找不到活動\n" + available_activities()

    return f"📋 {activity} 目前剩餘 {remaining} 個名額"


# 統計所有活動報名狀況
def statistics_activity():

    conn = sqlite3.connect(DB_NAME)

    try:
        query = """
            SELECT
                a.activity_name AS 活動名稱,
                a.activity_date AS 活動日期,
                a.capacity AS 總名額,
                COALESCE(
                    SUM(
                        CASE
                            WHEN r.status = '報名中'
                            THEN r.people_count
                            ELSE 0
                        END
                    ), 0
                ) AS 已報名人數,
                a.capacity - COALESCE(
                    SUM(
                        CASE
                            WHEN r.status = '報名中'
                            THEN r.people_count
                            ELSE 0
                        END
                    ), 0
                ) AS 剩餘名額

            FROM activities a

            LEFT JOIN registrations r
                ON a.activity_id = r.activity_id

            GROUP BY
                a.activity_id,
                a.activity_name,
                a.activity_date,
                a.capacity

            ORDER BY a.activity_id
        """

        df = pd.read_sql_query(query, conn)

        result = "📊 各活動報名統計：\n\n"

        for _, row in df.iterrows():
            result += (
                f"活動：{row['活動名稱']}\n"
                f"日期：{row['活動日期']}\n"
                f"總名額：{row['總名額']} 人\n"
                f"已報名：{row['已報名人數']} 人\n"
                f"剩餘名額：{row['剩餘名額']} 人\n"
                "--------------------\n"
            )

        return result

    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================

# 主流程：根據 AI 解讀結果執行對應功能
def main_process(user_text, student_name=""):

    # 檢查輸入是否為空白
    user_text = str(user_text).strip()
    student_name = str(student_name).strip()

    if not user_text:
        return "❌ 請輸入需求"

    # 呼叫第 4 格的 AI 解讀函式
    action, activity, people_count, raw = ai_interpret(user_text)

    # 統計所有活動
    if action == "統計":
        return statistics_activity()

    # 無法判斷動作或活動
    if action == "未知" or activity == "未知":
        return (
            "❌ 無法辨識需求或活動\n"
            + available_activities()
        )

    # 報名活動
    if action == "報名":
        return register_activity(
            student_name,
            activity,
            people_count
        )

    # 取消報名
    elif action == "取消":
        return cancel_activity(
            student_name,
            activity
        )

    # 查詢剩餘名額
    elif action == "查詢":
        return query_activity(activity)

    # 其他無法辨識的情況
    else:
        return "❌ 無法辨識需求\n" + available_activities()


# ============================================================
# 流程測試：依照規格書 7-2，共 10 步
# ============================================================

flow_test_cases = [
    ("小明", "我要報名迎新晚會，我和兩個朋友一起"),
    ("小明", "我要報名迎新晚會"),
    ("小華", "我要報名迎新晚會，總共三人"),
    ("小美", "我要報名迎新晚會"),
    ("小明", "我要報名程式工作坊"),
    ("小明", "我要報名籃球友誼賽"),
    ("小明", "幫我取消迎新晚會"),
    ("小美", "我要報名迎新晚會，總共兩人"),
    ("小明", "我要報名籃球友誼賽"),
    ("幹部", "各活動報名狀況")
]

# 每次測試前重建資料庫，確保結果一致
init_db()

print("========== 流程測試開始 ==========")

# 依序執行 10 個步驟
for i, (student_name, user_text) in enumerate(flow_test_cases, start=1):

    print(f"\n===== 第 {i} 步 =====")
    print("姓名：", student_name)
    print("使用者輸入：", user_text)

    # 執行主流程
    result = main_process(user_text, student_name)

    print("系統回應：")
    print(result)

print("\n========== 流程測試結束 ==========")
# ============================================================
# 手動輸入：讓使用者自行輸入姓名與需求
# ============================================================

student_name = input("請輸入姓名：")
user_text = input("請輸入需求：")

result = main_process(user_text, student_name)

print("\n===== 系統回應 =====")
print(result)

print("\n===== 最新資料表 =====")
show_tables()
# 最後顯示資料表
````
